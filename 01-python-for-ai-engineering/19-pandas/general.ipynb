# 🐼 Pandas — From Basic to Advanced

A comprehensive guide covering **Series**, **DataFrames**, **indexing**, **data cleaning**, **merging**, **groupby**, **time series**, **performance**, and more.

```python
import pandas as pd
import numpy as np

print(f'pandas version: {pd.__version__}')
print(f'numpy  version: {np.__version__}')
```

---
## 🟢 Level 1 — Basics

### 1.1 Series

```python
# From list
s = pd.Series([10, 20, 30], index=['a', 'b', 'c'], name='scores')
print(s)
print('dtype :', s.dtype)
print('index :', s.index.tolist())
print('values:', s.values)
```

```python
# From dict
s2 = pd.Series({'python': 90, 'sql': 85, 'ml': 92})
print(s2)
print('Max skill:', s2.idxmax(), '→', s2.max())
```

### 1.2 DataFrame Creation

```python
data = {
    'name' : ['Alice', 'Bob', 'Carol', 'Dave', 'Eve'],
    'dept' : ['Eng', 'DS', 'Eng', 'DS', 'PM'],
    'salary': [90000, 85000, 95000, 88000, 78000],
    'yoe'  : [5, 3, 7, 4, 2],
    'remote': [True, False, True, True, False],
}

df = pd.DataFrame(data)
df
```

```python
# Basic info
print(df.shape)      # (rows, cols)
print(df.dtypes)
df.info()
```

```python
df.describe()
```

### 1.3 Selecting Columns & Rows

```python
# Single column → Series
print(df['name'])

# Multiple columns → DataFrame
df[['name', 'salary']]
```

```python
# loc  — label-based
print(df.loc[0])          # row 0
print(df.loc[1:3, 'name':'salary'])  # slice rows & cols
```

```python
# iloc — position-based
print(df.iloc[0])         # first row
print(df.iloc[:, :3])     # all rows, first 3 cols
```

### 1.4 Basic Filtering

```python
# Boolean mask
df[df['salary'] > 88000]
```

```python
# Multiple conditions
df[(df['dept'] == 'Eng') & (df['yoe'] >= 5)]
```

```python
# query() — SQL-like syntax
df.query("dept == 'DS' and salary > 85000")
```

### 1.5 Adding & Modifying Columns

```python
df['salary_lpa'] = df['salary'] / 1000          # new column
df['senior']     = df['yoe'] >= 5               # boolean column
df['level']      = df['yoe'].apply(lambda y: 'Senior' if y >= 5 else 'Junior')
df
```

## 🟡 Level 2 — Intermediate

### 2.1 Sorting

```python
df.sort_values('salary', ascending=False)
```

```python
# Sort by multiple keys
df.sort_values(['dept', 'salary'], ascending=[True, False])
```

### 2.2 GroupBy & Aggregation

```python
# iterate over group
grouped = df.groupby('dept')
for group in grouped:
    print(group)
    print("-"*20)
```

```python
# Mean salary per department
df.groupby('dept')['salary'].mean()
```

```python
# Multiple aggregations
df.groupby('dept').agg(
    avg_salary=('salary', 'mean'),
    max_salary=('salary', 'max'),
    headcount =('name', 'count'),
    avg_yoe   =('yoe', 'mean'),
).round(2)
```

```python
# Group + filter (having equivalent)
dept_stats = df.groupby('dept')['salary'].mean()
dept_stats[dept_stats > 87000]
```

### 2.3 Missing Data

```python
# Introduce NaNs for demo
df_nan = df.copy()
df_nan.loc[1, 'salary'] = np.nan
df_nan.loc[3, 'dept']   = np.nan

print('NaN count per column:')
print(df_nan.isnull().sum())
print('\nAny NaN row:')
df_nan[df_nan.isnull().any(axis=1)]
```

```python
# Fill strategies
df_filled = df_nan.copy()
df_filled['salary'] = df_filled['salary'].fillna(df_filled['salary'].median())
df_filled['dept']   = df_filled['dept'].fillna('Unknown')
df_filled
```

```python
# Forward / backward fill (useful for time series)
ts = pd.Series([1, np.nan, np.nan, 4, np.nan])
print('ffill:', ts.ffill().tolist())
print('bfill:', ts.bfill().tolist())

# Drop rows where salary is NaN
df_nan.dropna(subset=['salary'])
```

### 2.4 Merging & Joining

```python
projects = pd.DataFrame({
    'name'   : ['Alice', 'Bob', 'Carol', 'Frank'],
    'project': ['Search', 'Recommend', 'Search', 'Ads'],
})

# inner join (default)
inner = df.merge(projects, on='name')
print('INNER:\n', inner[['name','dept','project']])

# left join
left = df.merge(projects, on='name', how='left')
print('\nLEFT:\n', left[['name','dept','project']])
```

```python
# concat — stack DataFrames
df_new = pd.DataFrame({'name':['Frank'],'dept':['PM'],'salary':[72000],'yoe':[1],'remote':[True]})
df_all = pd.concat([df[['name','dept','salary','yoe','remote']], df_new], ignore_index=True)
df_all
```

### 2.5 String Operations

```python
s = pd.Series(['Alice Smith', 'bob jones', '  Carol White  ', 'DAVE BROWN'])

print(s.str.lower())
print("-"*20)
print(s.str.upper())
print("-"*20)
print(s.str.strip())
print("-"*20)
print(s.str.title())
print("-"*20)
print(s.str.contains('alice', case=False))
print("-"*20)
print(s.str.split(' ').str[0])     # first name
print("-"*20)
print(s.str.replace(r'\s+', '_', regex=True))
```

### 2.6 apply / map / transform

```python
# apply on Series
df['salary_band'] = df['salary'].apply(lambda x: 'High' if x > 90000 else ('Mid' if x > 82000 else 'Low'))
print(df[['name','salary','salary_band']])
print("-"*20)
# map — element-wise (replace values)
dept_map = {'Eng': 'Engineering', 'DS': 'Data Science', 'PM': 'Product'}
df['dept_full'] = df['dept'].map(dept_map)
print(df[['dept','dept_full']])
```

```python
# transform — keeps original index shape (useful with groupby)
df['dept_avg_salary'] = df.groupby('dept')['salary'].transform('mean')
df['salary_vs_avg']   = df['salary'] - df['dept_avg_salary']
df[['name','dept','salary','dept_avg_salary','salary_vs_avg']]
```

### 3.3 Time Series

```python
# Create a date-indexed series
dates = pd.date_range('2024-01-01', periods=12, freq='ME')  # month-end
revenue = pd.Series(
    np.random.randint(100_000, 500_000, size=12),
    index=dates,
    name='revenue'
)
revenue
```

```python
# Resample to quarterly sum
print('Quarterly revenue:')
print(revenue.resample('QE').sum())

# Rolling 3-month average
print('\n3-month rolling avg:')
print(revenue.rolling(3).mean().round(0))
```

```python
# Date component extraction
ts_df = pd.DataFrame({'date': pd.date_range('2024-01-01', periods=6, freq='D'), 'value': range(6)})
ts_df['year']    = ts_df['date'].dt.year
ts_df['month']   = ts_df['date'].dt.month
ts_df['weekday'] = ts_df['date'].dt.day_name()
ts_df
```

### 3.4 Window Functions

```python
# Rank within group (like SQL RANK OVER PARTITION BY)
df['rank_in_dept'] = df.groupby('dept')['salary'].rank(ascending=False, method='dense').astype(int)
df[['name','dept','salary','rank_in_dept']].sort_values(['dept','rank_in_dept'])
```

```python
# Cumulative sum
df_sorted = df.sort_values('salary', ascending=False).reset_index(drop=True)
df_sorted['cumsum_salary'] = df_sorted['salary'].cumsum()
df_sorted[['name','salary','cumsum_salary']]
```

### 3.5 Melt & Stack/Unstack (Reshaping)

```python
wide = pd.DataFrame({
    'name': ['Alice','Bob'],
    'q1': [100, 80],
    'q2': [120, 90],
    'q3': [110, 95],
})

# wide → long
long = wide.melt(id_vars='name', var_name='quarter', value_name='revenue')
print(long)
print("-"*20)
# long → wide
back = long.pivot(index='name', columns='quarter', values='revenue')
back
```

### 3.6 Categorical Dtype & Memory Optimization

```python
before = df['dept'].memory_usage(deep=True)
df['dept_cat'] = df['dept'].astype('category')
after  = df['dept_cat'].memory_usage(deep=True)
print(f'dept  object:   {before} bytes')
print("-"*20)
print(f'dept  category: {after} bytes')
print("-"*20)
print(f'Reduction: {(1 - after/before)*100:.1f}%')
print("-"*20)
print(df['dept_cat'].cat.categories.tolist())
```

### 3.7 Custom Aggregation with pipe

```python
def add_senior_flag(df_):
    return df_.assign(is_senior=df_['yoe'] >= 5)

def normalize_salary(df_):
    return df_.assign(salary_norm=(df_['salary'] - df_['salary'].min()) / (df_['salary'].max() - df_['salary'].min()))

result = (
    df[['name','dept','salary','yoe']]
    .pipe(add_senior_flag)
    .pipe(normalize_salary)
    .sort_values('salary_norm', ascending=False)
)
result
```

### 3.9 Advanced GroupBy — Named Aggregation & Filter

```python
# GroupBy filter — keep only groups with > 1 member
df_filtered = df.groupby('dept').filter(lambda g: len(g) > 1)
print('Depts with >1 member:')
print(df_filtered[['name','dept']])
```

```python
# GroupBy apply — custom function per group
def top_earner(group):
    return group.nlargest(1, 'salary')[['name','salary']]

df.groupby('dept', group_keys=False).apply(top_earner)
```

### 3.10 Performance — vectorized ops vs apply

```python
import time

big = pd.DataFrame({'x': np.random.randn(500_000)})

# apply (slow)
t0 = time.perf_counter()
big['x'].apply(lambda v: v ** 2)
t_apply = time.perf_counter() - t0
print("-"*20)
# vectorized (fast)
t0 = time.perf_counter()
big['x'] ** 2
t_vec = time.perf_counter() - t0

print(f'apply     : {t_apply*1000:.1f} ms')
print("-"*20)
print(f'vectorized: {t_vec*1000:.1f} ms')
print("-"*20)
print(f'Speedup   : {t_apply/t_vec:.0f}x')
```

### 3.12 Method Chaining (Fluent Style)

```python
result = (
    df
    .query("dept != 'PM'")
    .assign(bonus=lambda d: d['salary'] * 0.10)
    .groupby('dept')
    .agg(total_bonus=('bonus', 'sum'), headcount=('name', 'count'))
    .sort_values('total_bonus', ascending=False)
    .reset_index()
)
result
```

## ✅ Quick Reference Cheatsheet

| Operation | Code |
|-----------|------|
| Select column | `df['col']` |
| Filter rows | `df[df['col'] > val]` / `df.query(...)` |
| Label-based select | `df.loc[rows, cols]` |
| Position-based select | `df.iloc[r, c]` |
| Group + aggregate | `df.groupby('col').agg(...)` |
| Merge | `df.merge(df2, on='key', how='left')` |
| Stack vertically | `pd.concat([df1, df2])` |
| Pivot | `df.pivot_table(values, index, columns, aggfunc)` |
| Reshape wide→long | `df.melt(id_vars, var_name, value_name)` |
| Fill NaN | `df.fillna(val)` / `ffill()` / `bfill()` |
| Drop NaN | `df.dropna(subset=['col'])` |
| Rank in group | `df.groupby('g')['col'].rank()` |
| Rolling window | `s.rolling(n).mean()` |
| Resample time | `s.resample('QE').sum()` |
| Chain methods | `df.pipe(fn1).pipe(fn2)` |
